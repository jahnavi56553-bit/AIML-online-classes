```python
import pandas as pd
df = pd.read_csv("train_u6lujuX_CVtuZ9i (1).csv")
df
```

```python
df.columns
```

```python
df.rename(columns={
    "Married" : "Marital_status"
},inplace=True)
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender']=='f')
```

```python
df['Dependents'] = df['Dependents'].fillna(df['Dependents']==0)
```

```python
df['Self_Employed'] = df['Self_Employed'].fillna(df['Self_Employed']=="Yes")
```

```python
df['LoanAmount'] = df['LoanAmount'].fillna(df['LoanAmount']==0)
```

```python
df['Loan_Amount_Term'] = df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term']==180)
```

```python
df['Credit_History'] = df['Credit_History'].fillna(df['Credit_History']==0)
```

```python
df['Marital_status'] = df['Marital_status'].fillna(df['Marital_status']=="Yes")
```

```python
df.sort_values("ApplicantIncome").tail(10)
```

```python
df.drop_duplicates()
```

```python
q1 = df['ApplicantIncome'].quantile(0.25)
q3 = df['ApplicantIncome'].quantile(0.75)

IQR = q3 - q1

lower_limit = q1 - 1.5 * IQR
upper_limit = q3 + 1.5 * IQR
print("Q1: ",q1)
print("Q3: ",q3)
print("IQR: ",IQR)
print("Lower: ",lower_limit)
print("Upper: ",upper_limit)
outliers = df[
    (df["ApplicantIncome"]<lower_limit) |
    (df["ApplicantIncome"]>upper_limit)
]
print("\nNumber of ApplicantIncome outliers: ",len(outliers))
print("\nApplicantIncome outliers:")
print(outliers[["ApplicantIncome","LoanAmount"]].head(20))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,5))
plt.boxplot(df["ApplicantIncome"])
plt.title("ApplicantIncome before outlier treatment")
plt.ylabel('ApplicantIncome')
plt.show()
```

```python
df['ApplicantIncome_Updated'] = df['ApplicantIncome'].clip(
    lower = lower_limit,
    upper = upper_limit
)
```

```python

plt.figure(figsize=(6,5))
plt.boxplot(df["ApplicantIncome_Updated"])
plt.title("ApplicantIncome after outlier treatment")
plt.ylabel('ApplicantIncome')
plt.show()
```

```python
Average = df.groupby("Property_Area")['ApplicantIncome'].mean()
plt.figure(figsize=(7,5))
plt.bar(Average.index,Average.values,color=["red","yellow","green"])
plt.title("Average of ApplicantIncome by Property_Area")
plt.xlabel = ("Property_Area")
plt.ylabel = ("ApplicantIncome")
plt.show()
```

```python
Loan_Amount_Term_vs_LoanAmount = df.groupby("Loan_Amount_Term")["LoanAmount"].mean()
print("\nAverage LoanAmount by Loan_Amount_Term:")
print(Loan_Amount_Term_vs_LoanAmount)
plt.figure(figsize=(10,5))
plt.plot(
    Loan_Amount_Term_vs_LoanAmount.index,
    Loan_Amount_Term_vs_LoanAmount.values,
    marker="."
)

plt.title("Average LoanAmount by Loan_Amount_Term ")
plt.xlabel = ("Loan_Amount_Term")
plt.ylabel = ("LoanAmount")
plt.grid(True)
plt.show()
```
