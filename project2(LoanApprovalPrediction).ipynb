```python
q1 = df['ApplicantIncome'].quantile(0.25)
q3 = df['ApplicantIncome'].quantile(0.75)

IQR = q3 - q1

lower_limit = q1 - 1.5 * IQR
upper_limit = q3 + 1.5 * IQR
print("Q1: ",q1)
print("Q3: ",q3)
print("IQR: ",IQR)
print("Lower: ",lower_limit)
print("Upper: ",upper_limit)
outliers = df[
    (df["ApplicantIncome"]<lower_limit) |
    (df["ApplicantIncome"]>upper_limit)
]
print("\nNumber of ApplicantIncome outliers: ",len(outliers))
print("\nApplicantIncome outliers:")
print(outliers[["ApplicantIncome","LoanAmount"]].head(20))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,5))
plt.boxplot(df["ApplicantIncome"])
plt.title("ApplicantIncome before outlier treatment")
plt.ylabel('ApplicantIncome')
plt.show()
```

```python
df['ApplicantIncome_Updated'] = df['ApplicantIncome'].clip(
    lower = lower_limit,
    upper = upper_limit
)
```

```python

plt.figure(figsize=(6,5))
plt.boxplot(df["ApplicantIncome_Updated"])
plt.title("ApplicantIncome after outlier treatment")
plt.ylabel('ApplicantIncome')
plt.show()
```

```python
import numpy as np
X = df['LoanAmount']
Y = df['ApplicantIncome']
m = ((X - X.mean())*(Y - Y.mean()))/((X - X.mean())**2).sum()
c=Y.mean() - m*X.mean()
Y_pred = m*X+c
print(Y_pred)
Y_pred = np.round(Y_pred).astype(int)
print(Y_pred.head())
print("Slope: ",m)
print("Intercept: ",c)

```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.scatter(X,Y)
plt.plot(X,Y_pred,color="red")
plt.title("simple linear chart")
plt.xlabel("LoanAmount")
plt.ylabel("ApplicantIncome")
plt.show()
```

```python
import matplotlib.pyplot as plt
sorted_data = pd.DataFrame({
    "X":X,
    "Actual" : Y,
    "Predicted":Y_pred
}).sort_values("X")
plt.figure(figsize=(8,5))
plt.plot(sorted_data["X"],sorted_data["Actual"],label="Actual LoanAmount")
plt.plot(sorted_data["X"],sorted_data["Predicted"],label="Predicted loanAmount")
plt.title("Actual vs Predicted LoanAmount")
plt.xlabel("LoanAmount")
plt.ylabel("LoanAmount")
plt.legend()
plt.grid(True)
plt.show()
```
